# Práctica 5: Descenso por Gradiente Vectorizado con Gráficas de Costo

## 1. Fundamentos Teóricos
Vectorizamos el cálculo del error cuadrático medio (MSE) y del gradiente con respecto a $w$.

### Formulación Vectorizada
* **Errores Cuadráticos:** $\vec{E} = (\vec{y} - \vec{t})^2$
* **Costo (MSE):** $J = \text{np.mean}(\vec{E})$
* **Gradientes por Muestra:** $\frac{\partial E}{\partial w} = 2(\vec{y} - \vec{t}) \cdot \vec{x}$
* **Gradiente Global:** $\nabla_w J = \text{np.mean}\left(\frac{\partial E}{\partial w}\right)$

```python
import time
import numpy as np
import matplotlib.pyplot as plt

inputs = np.array([1, 2, 3, 4], dtype=float)
targets = np.array([2, 4, 6, 8], dtype=float)

w = 0.1
epochs = 10
learning_rate = 0.1
pause_time = 0

weight_history = []
cost_history = []

for epoch in range(epochs):
    predictions = w * inputs
    errors = (predictions - targets) ** 2
    cost = np.mean(errors)
    
    errors_d = 2 * (predictions - targets)
    weight_d = errors_d * inputs
    gradient = np.mean(weight_d)
    
    w -= learning_rate * gradient
    
    weight_history.append(w)
    cost_history.append(cost)
    time.sleep(pause_time)

plt.figure(figsize=(8, 4))
plt.plot(range(1, epochs + 1), cost_history, marker="o", color="purple")
plt.xlabel("Época")
plt.ylabel("Costo J")
plt.title("Evolución de la Función de Costo J")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 4))
plt.plot(range(1, epochs + 1), weight_history, marker="o", color="green")
plt.xlabel("Época")
plt.ylabel("Peso w")
plt.title("Ajuste del Peso w por Descenso por Gradiente")
plt.grid(True)
plt.show()
```

## 2. Interpretación de Gráficas
* **Curva de Costo:** Se observa una caída exponencial drástica de la función de costo $J$, tendiendo a 0 en menos de 10 épocas.
* **Oscilación del Peso:** Dependiendo de la tasa de aprendizaje, se pueden observar ligeras oscilaciones iniciales que rápidamente se corrigen hacia $w = 2.0$.
