# ***3. Eliminar nulos y crear una nueva tabla***


La herramienta `.dropna()` funciona como una aspiradora: detecta cualquier fila que tenga un dato vacío y la elimina.
* `nbaNoNull = nba.dropna()`: En lugar de modificar nuestra tabla original, aplicamos la "aspiradora" y guardamos el resultado limpio dentro de una **nueva variable** llamada `nbaNoNull`. Así conservamos intacta la versión original por si la necesitamos después.

```python
nbaNoNull = nba.dropna()
nbaNoNull
```

# ***4. Simulacro de limpieza vs Guardado permanente***


Aquí reforzamos la lección más importante de Pandas sobre borrado de datos:
* `.dropna()`: Si lo ejecutamos solo, nos muestra una vista previa (simulacro) de cómo se vería la tabla limpia.
* `nba`: Al imprimir la tabla inmediatamente después, vemos que los datos vacíos siguen ahí.
* `#nba.dropna(inplace=True)`: El código inactivo demuestra que, al igual que con el método `.drop()`, necesitamos usar `inplace=True` si queremos que la "aspiradora" elimine los nulos permanentemente en la memoria.

```python
#nba.dropna(inplace=True)
nba.dropna()
nba
```

# ***5. Revisar el estado de la tabla***


Verificamos el estado técnico de nuestros datos en la memoria:
* `.info()`: Nos recuerda cuántas filas, columnas y datos válidos tenemos disponibles actualmente en la tabla original `nba`.

```python
nba.info()
nba
```

# ***6. Eliminar filas completamente vacías***


A veces no queremos borrar una fila si solo le falta un dato, sino únicamente cuando está totalmente en blanco:
* `how="all"`: Al agregar este parámetro dentro de la aspiradora `.dropna()`, le decimos a Pandas que sea más compasivo: "solo borra esta fila si **absolutamente todas** sus columnas están vacías".

```python
nba.dropna(how="all")
```

# ***7. Decidir qué dirección limpiar (Ejes)***


Por defecto, la aspiradora borra filas (horizontal), pero podemos cambiar su comportamiento:
* `axis = 0`: Es el valor por defecto. Le indica a Pandas que elimine las **filas** enteras que contengan algún dato vacío.
* `axis = 1`: Cambia la dirección de limpieza a vertical. Le indica a Pandas que elimine **columnas completas** si detecta que falta al menos un solo dato en ellas. (Es una limpieza muy agresiva).

```python
nba.dropna(axis = 0)
```

```python
nba.dropna(axis = 1)
```

# ***8. Limpieza enfocada en columnas clave***


Esta es la técnica más profesional. Nos permite dictar exactamente en dónde importa que falten datos:
* `subset = [ ]`: Le pasamos una lista de Python con nombres de columnas. Esto le dice a Pandas: "revisa únicamente las columnas Name, Salary y College. Si falta un dato en alguna de estas tres, borra la fila. Si el dato falta en otra columna diferente, ignóralo y no borres nada".

```python
nba.dropna(subset = ["Name", "Salary", "College"])
```

# ***9. Resumen del proceso completo***


En conclusión, este archivo nos enseña el arsenal completo de herramientas para purificar nuestra base de datos manejando los valores nulos:
* **Seguridad de la información:** Aprendimos que en lugar de usar `inplace=True`, podemos guardar nuestra limpieza en una nueva variable (`nbaNoNull`) para proteger los datos originales.
* **Flexibilidad de limpieza:** Comprendimos cómo modificar el comportamiento estricto del borrado usando parámetros como `how="all"` (borrar solo filas 100% vacías) o `axis=1` (borrar columnas enteras en lugar de filas).
* **Limpieza inteligente:** Descubrimos que el parámetro `subset` nos permite realizar limpiezas quirúrgicas, salvando información valiosa al enfocar la "aspiradora" únicamente en las columnas que consideramos vitales para nuestro análisis.
