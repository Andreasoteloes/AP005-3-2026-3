# **2. Cargar datos y explorar categorías únicas**


* `read_csv()`: Como este archivo es de tipo texto separado por comas (`.csv`), usamos esta función en lugar de `read_excel`.
* `.unique()`: Nos muestra una lista con todos los valores diferentes que existen dentro de una columna (sin repetir), ideal para ver las categorías disponibles (ej. géneros).

```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/aplicada/restaurant.csv")
df
df.info()
df["Gender"].unique()
```

# **3. Crear tablas dinámicas (pivot_table)**


Una tabla dinámica resume y agrupa datos automáticamente, igual que en Excel:
* `values = "Amount"`: La columna numérica que queremos calcular o sumar.
* `index = "Gender"`: La categoría que usaremos para organizar las filas.
* `aggfunc = "sum"` / `"mean"`: La función matemática a aplicar (suma total o promedio).

```python
df.pivot_table(values = "Amount", index = "Gender", aggfunc="sum")
df.pivot_table(values = "Amount", index = "Gender", aggfunc="mean")
df.pivot_table(values = "Amount", index = "Menu", aggfunc="sum")
```

# **4. Tablas dinámicas avanzadas (Múltiples índices y columnas)**


Podemos cruzar varias variables para obtener reportes más detallados:
* `index = ["Gender", "Menu"]`: Crea una jerarquía de filas agrupando por género y luego por menú.
* `columns = "City"` o `columns = ["Frequency", "City"]`: Distribuye la información horizontalmente separándola por columnas adicionales.

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], aggfunc="sum")
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = "City", aggfunc="sum")
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = ["Frequency", "City"], aggfunc="sum")
```
