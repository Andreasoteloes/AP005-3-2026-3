# **3. Ordenar datos alfabéticamente**


Podemos organizar los registros basándonos en columnas de texto:
* `.sort_values("Name")`: Toma la columna de nombres y reorganiza toda la tabla de la A a la Z.
* `.head(20)`: Como una tabla de la NBA puede ser muy larga, encadenamos este método al final para ver únicamente los primeros 20 resultados de nuestro listado ya ordenado.

```python
nba.sort_values("Name").head(20)
```

# **4. Orden numérico y parámetros de control**


Podemos aplicar el mismo principio a columnas numéricas, como los salarios:
* `sort_values("Salary")`: Por defecto, Pandas siempre organiza los números de menor a ascendente (de menor a mayor).
* `#nba.sort_values(..., ascending = False, inplace = True)`: El código inactivo nos muestra una combinación muy potente. Si cambiamos `ascending` a `Falso`, la tabla se ordena de **mayor a menor** (ideal para ver los salarios más altos primero), y el `inplace = True` guardaría ese orden permanentemente en la memoria.

```python
nba.sort_values("Salary")
#nba.sort_values("Salary", ascending = False, inplace = True)
```

# **5. Orden jerárquico (Múltiples columnas) y control de nulo**s


A veces necesitamos organizar la información usando más de un criterio a la vez:
* `["Team", "Name"]`: Al pasar una lista con varias columnas, Pandas ordena primero por equipo. Si encuentra jugadores en el mismo equipo, los ordena alfabéticamente por su nombre (como una jerarquía de orden).
* `#na_position = 'first'`: El código comentado nos enseña un truco avanzado. Por defecto, los datos vacíos (`NaN`) se van al final de la tabla; este parámetro nos permite decidir si queremos que los huecos aparezcan obligatoriamente al principio (`'first'`).

```python
nba.sort_values(["Team", "Name"])
#nba.sort_values(["Team", "Name"], na_position = 'first')
```
