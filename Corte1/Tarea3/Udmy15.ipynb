# **2. Cargar datos y ordenar por contenido (Values)**


Importamos nuestra librería y aplicamos un ordenamiento avanzado sobre las columnas:
* `import pandas as pd`: Llamamos a Pandas.
* `read_excel()`: Cargamos el archivo de la NBA desde la ruta exacta del Drive.
* `sort_values(["Team", "Name"])`: Ordena la tabla jerárquicamente primero por equipo y luego por nombre.
* `na_position = "first"`: Envía obligatoriamente todas las celdas vacías (`NaN`) al principio de la lista.
* `inplace = True`: Guarda esta reorganización de forma permanente en la tabla.

```python
import pandas as pd
nba = pd.read_excel('/content/drive/MyDrive/aplicada/Nba Teams.xlsx')
nba.sort_values(["Team", "Name"], na_position = "first", inplace = True)
nba
```

# **3. Ordenar por índice original (sort_index)**


Cuando aplicamos un ordenamiento por valores (como hicimos en el bloque anterior), los números de las filas (el índice del lado izquierdo) quedan completamente revueltos y salteados. Para solucionar esto usamos:
* `.sort_index()`: Esta herramienta inspecciona las etiquetas de las filas y devuelve la tabla a su orden numérico secuencial original (0, 1, 2, 3...), deshaciendo el caos que causó el filtro anterior o recuperando la estructura inicial de cuando cargamos el archivo.

```python
nba.sort_index()
```

# **4. Resumen del proceso completo**


En conclusión, este archivo nos enseña a manejar la anatomía y las etiquetas de las filas en Pandas:
* **El impacto del ordenamiento:** Comprendimos que al ordenar una tabla por sus valores, las etiquetas numéricas de las filas pierden su secuencia natural.
* **Restauración de la tabla:** Descubrimos `.sort_index()`, el método clave para restablecer el orden original basado en los índices, asegurando que podamos alternar entre vistas ordenadas por contenido y vistas ordenadas por posición cada vez que lo necesitemos.
