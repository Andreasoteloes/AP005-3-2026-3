# ***2. Cargar la tabla de datos***


Como en todo análisis, el primer paso es abrir nuestro archivo para poder trabajar con él:
* `import pandas as pd`: Traemos nuestra librería de análisis.
* `read_excel()`: Cargamos el archivo local que contiene los datos de los equipos de la NBA.
* `.head()`: Imprimimos las primeras 5 filas para comprobar visualmente los nombres exactos de las columnas que vamos a utilizar.

```python
import pandas as pd
nba = pd.read_excel("/content/drive/MyDrive/Colab Notebooks/udmy/Nba Teams.xlsx")
nba.head()
```

# ***3. Extraer una columna específica (Método de corchetes)***


Para analizar una sola variable sin distraernos con el resto de la tabla, podemos "arrancar" una sola columna:
* `["Name"]`: Usamos el nombre de la variable de nuestra tabla (`nba`) seguido de corchetes y comillas. Adentro escribimos el nombre exacto de la columna. En este caso, nos traerá únicamente la lista con los nombres de los jugadores.

```python
nba["Name"]
```

# ***4. Extraer columnas numéricas***


Podemos aplicar exactamente la misma técnica para cualquier tipo de dato:
* `["Salary"]`: Al usar nuevamente los corchetes con el nombre exacto de la columna, extraemos ahora únicamente la información financiera (los salarios). Esta es la forma más segura de extraer datos en Pandas.

```python
nba["Salary"]
```

# ***5. Extraer una columna (Notación de punto)***


Existe una segunda forma de hacer exactamente lo mismo, que es más rápida para escribir:
* `.Name`: Usamos la variable de nuestra tabla (`nba`) seguida de un punto y el nombre de la columna, sin usar corchetes ni comillas. Hace exactamente lo mismo que `nba["Name"]`.
* **La regla de oro:** Este método rápido solo funciona si el nombre de la columna es una sola palabra. Si tuviera espacios (ejemplo: "First Name"), este método daría error y estaríamos obligados a usar los corchetes.

```python
nba.Name
```

En conclusión, este archivo nos enseña a enfocarnos en variables específicas de nuestra base de datos:

* **Enfoque analítico:** Aprendimos que podemos aislar columnas individuales (Series) para analizarlas por separado, sin cargar toda la tabla.
* **Dos caminos, un resultado:** Descubrimos los dos métodos de extracción: la *notación de corchetes* (`nba["Name"]`) y la *notación de punto* (`nba.Name`).
* **Seguridad vs. Rapidez:** Entendimos que aunque el método del punto es más rápido, el uso de corchetes es la práctica más segura a prueba de errores, especialmente si los nombres de las columnas contienen espacios.
