Una vez conectados, preparamos nuestro espacio de trabajo:

* `read_excel()`: Cargamos el archivo con los datos completos de la NBA.

# ***3. Extraer múltiples columnas (Doble corchete)***


Para aislar más de una variable a la vez, ya no nos sirve el método del punto ni un solo corchete. Necesitamos usar **dobles corchetes**:
* `[[ ]]`: El primer par de corchetes (los de afuera) le indican a Pandas "voy a extraer algo". El segundo par (los de adentro) forman una **lista tradicional de Python** con los nombres de lo que queremos extraer.
* `["Name", "Team"]`: Al pasar esta lista, Pandas nos devuelve una "mini-tabla" que contiene únicamente la columna de nombres y la de equipos.

```python
nba[["Name", "Team"]]
```

# ***4. Expandir nuestra tabla filtrada***


Podemos agregar tantas columnas como necesitemos dentro de nuestra lista interna, simplemente separándolas por comas:
* `[["Name", "Team", "Salary"]]`: Le pasamos una lista de tres elementos. El resultado es una tabla perfectamente limpia que excluye cualquier otra información (como el peso o la edad) y nos deja solo con los nombres, equipos y salarios.

```python
nba[["Name", "Team", "Salary"]]
```

# ***5. Resumen del proceso completo***


En conclusión, este archivo nos enseña a crear sub-tablas o vistas filtradas de nuestra información original:
* **La lógica del doble corchete `[[ ]]`:** Comprendimos que para extraer múltiples columnas no basta con escribir sus nombres, sino que debemos agruparlas dentro de una lista de Python (el corchete interior).
* **Evolución de la extracción:** En el archivo anterior vimos que al extraer una sola columna obtenemos una "Serie" (una lista simple). Ahora descubrimos que al extraer dos o más columnas, el resultado que obtenemos es un nuevo "DataFrame" (una tabla en miniatura), ideal para exportar o presentar informes enfocados en datos específicos.
