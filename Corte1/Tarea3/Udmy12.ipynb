# **2. Cargar el archivo de Excel**


Importamos las herramientas necesarias y abrimos nuestra base de datos:
* `import pandas as pd`: Llamamos a la librería Pandas.
* `read_excel()`: Cargamos el archivo de la NBA usando su ruta exacta en el Drive.

```python
import pandas as pd
nba = pd.read_excel('/content/drive/MyDrive/aplicada/Nba Teams.xlsx')
```

# **3. Aislar la columna de salarios**


Antes de hacer matemáticas, recordamos cómo extraer una sola variable:
* `nba["Salary"]`: Extraemos únicamente la columna que contiene los salarios de los jugadores para poder aplicarle operaciones numéricas.

```python
nba["Salary"]
```

# **4. Operaciones de Suma (Suma masiva)**


Podemos sumar un valor fijo a todos los registros de la columna al mismo tiempo. Existen dos formas de hacerlo:
* `.add(50000)`: El método formal de Pandas para sumar una cantidad a cada celda de la Serie.
* `+ 2000000`: El operador matemático tradicional de Python. Hace exactamente lo mismo y es mucho más rápido de escribir.

```python
nba["Salary"].add(50000)
```

```python
nba["Salary"] + 2000000
```

# **5. Crear una nueva columna basada en operaciones**


No solo podemos calcular valores al aire, sino guardarlos permanentemente en la tabla:
* `nba["bonus_salary"] = ...`: Al combinar la asignación que vimos en clases pasadas con una suma matemática (`+ 30000`), creamos una columna completamente nueva que calcula automáticamente el bono para cada jugador.

```python
nba["bonus_salary"] = nba["Salary"] + 30000
nba
```

# **6. Operaciones de Resta**


De igual forma, podemos restar cantidades a toda la columna de manera masiva:
* `.sub(55000)`: El método formal de Pandas para restar un valor a cada registro.
* `#nba["Salary"] - sub(55000)`: El profesor deja este código comentado como ejemplo de un error común de sintaxis que debemos evitar.

```python
nba["Salary"].sub(55000)
#nba["Salary"] - sub(55000)
```

#**7. Operaciones de División**


Útil para simplificar números grandes o convertir unidades:
* `.div(1000000)`: Divide cada salario de la columna entre un millón (ideal para simplificar reportes financieros).
* `/ 1000000`: Su equivalente usando el operador tradicional de barra diagonal, que cumple exactamente la misma función.

```python
nba["Salary"].div(1000000)
#nba["Salary"]/1000000
```

# **8. Operaciones de Multiplicación (Conversión de unidades)**


Un uso práctico excelente de las matemáticas en Pandas es convertir sistemas de medida a nivel masivo:
* `.mul(0.453592)` o `* 0.453592`: Multiplica la columna de peso (`Weight`) por este factor constante para convertir masivamente las libras a kilogramos en toda la tabla en un solo paso.

```python
nba["Weight"].mul(0.453592)
#nba["Weight"]*0.453592
```

# **9. Resumen del proceso completo**


En conclusión, este archivo nos enseña el poder del cálculo vectorial (matemáticas masivas) en Pandas:
* **Adiós a los bucles o celdas manuales:** Comprendimos que Pandas aplica operaciones aritméticas a columnas enteras de manera simultánea, ahorrando miles de líneas de código.
* **Métodos vs. Operadores:** Descubrimos que tenemos dos caminos para operar datos: usar métodos formales (como `.add()` o `.div()`) o usar los operadores matemáticos tradicionales de Python (`+`, `-`, `/`, `*`).
* **Creación de valor:** Aprendimos a reciclar las operaciones matemáticas para alimentar columnas completamente nuevas (como el `bonus_salary`), lo que nos permite enriquecer nuestras bases de datos originales.
