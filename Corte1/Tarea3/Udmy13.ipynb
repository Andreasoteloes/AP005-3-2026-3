# **3. Conteo de valores (value_counts)**


Si queremos saber cuántos jugadores pertenecen a cada equipo, no necesitamos contarlos uno por uno. Pandas tiene una herramienta automática para esto:
* `.value_counts()`: Este método revisa toda la columna extraída (`nba["Team"]`) y agrupa los datos. Nos devuelve un reporte exacto de cuántas veces se repite cada equipo en la tabla.
* **Orden por defecto:** De manera automática, Pandas siempre nos muestra los resultados ordenados de mayor a menor (descendente), es decir, el equipo con más jugadores aparecerá primero.

```python
nba["Team"].value_counts()
```

# **4. Modificar el orden de los resultados (ascending)**


Podemos aplicar el mismo conteo a otras columnas (como la división a la que pertenece el equipo) y cambiar la forma en que se muestran los resultados:
* `ascending = True`: Al agregar este parámetro dentro del paréntesis, obligamos a Pandas a invertir su regla por defecto. Ahora nos mostrará los resultados ordenados de **menor a mayor** (ascendente).
* El código inactivo (`#nba["Division"].value_counts()`) nos sirve de contraste para ver cómo se habría organizado la lista sin este parámetro especial.

```python
#nba["Division"].value_counts()
nba["Division"].value_counts(ascending = True)
```

# **5. Resumen del proceso completo**


En conclusión, este archivo nos enseña a agrupar y cuantificar nuestra información rápidamente:

* **Control del orden:** Aprendimos a manipular la presentación de nuestros resultados utilizando el parámetro `ascending = True`, lo cual es muy útil cuando buscamos encontrar los datos menos comunes o atípicos al principio de nuestra lista.
