# **2. Cargar los datos y revisar el estado general**


Importamos nuestra librería principal y abrimos el archivo:

```python
import pandas as pd
insta = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/udmy/aplicada/Instagram.csv')
insta.head()
insta.info()
```

# **3. Convertir columnas de texto a formato Fecha (Datetime**)


Por defecto, Pandas lee las fechas que vienen en archivos CSV como simples cadenas de texto (strings), lo que impide hacer cálculos con ellas. Para solucionar esto:
* `pd.to_datetime()`: Toma una columna específica (como la fecha de registro o el último inicio de sesión) y la transforma oficialmente en un objeto de tipo fecha y hora, permitiéndonos realizar operaciones temporales en el futuro.

```python
insta["reg date"] = pd.to_datetime(insta["reg date"])
insta["last login"] = pd.to_datetime(insta["last login"])
insta.info()
```

# **4. Optimizar memoria RAM usando tipos Categóricos**


En bases de datos gigantes, almacenar palabras repetitivas (como géneros o países) como texto común desperdicia muchísima memoria. Podemos optimizarlo:
* `.nunique()`: Cuenta cuántos valores únicos existen en una columna para verificar si es apta para convertirse en categoría (si tiene pocos valores repetidos muchas veces).
* `.astype("category")`: Convierte el formato de la columna. En lugar de guardar cada palabra completa repetida miles de veces, Pandas asigna códigos numéricos internos ocultos, reduciendo drásticamente el peso del archivo en la memoria del computador.

```python
insta["gender"].nunique()
insta["gender"] = insta["gender"].astype("category")
insta.info()
insta["Country"]
```
