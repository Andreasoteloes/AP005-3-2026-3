# ***4. Detectar celdas vacías (Datos nulos)***


En el mundo real, las bases de datos suelen tener huecos o celdas sin rellenar. Pandas nos ayuda a encontrarlas:
* `.isnull()`: Este método revisa celda por celda toda la tabla y responde con Falso o Verdadero (Valores Booleanos). Si responde `False`, significa que la celda tiene información. Si responde `True`, significa que la celda está completamente vacía (nula).

```python
nba.isnull()
```

# ***5. Conteo exacto de datos vacíos por columna***


Ver un mapa gigante de "Falso" y "Verdadero" es difícil de leer. Para solucionarlo, recurrimos al encadenamiento de métodos:
* `.isnull().sum()`: Primero aplicamos el método que detecta los vacíos (`.isnull()`). Inmediatamente después, le encadenamos una sumatoria matemática (`.sum()`).
* **La lógica matemática:** En Python, el valor booleano `True` equivale a 1. Al sumar todos los "Trues", Pandas nos entrega un reporte perfecto y fácil de leer con la cantidad exacta de datos que faltan por cada columna.

```python
nba.isnull().sum()
```
