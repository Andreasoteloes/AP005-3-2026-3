# ***1. Conectar Google Drive a Colab (Mounting)***


Como Colab es un entorno virtual en la nube, por defecto no tiene acceso a nuestros archivos personales. Para poder leer documentos (como archivos de Excel o CSV) que tenemos en nuestra cuenta, necesitamos establecer una conexión:
* `from google.colab import drive`: Llama a una herramienta especial y exclusiva de Google Colab diseñada para interactuar con nuestro almacenamiento.
* `drive.mount('/content/drive')`: Abre un "puente" o ruta entre nuestro cuaderno de Colab y nuestro Google Drive. Al ejecutarlo, nos pedirá permisos de seguridad y, una vez aceptados, nos permitirá leer y guardar archivos directamente en nuestra nube.

```python
from google.colab import drive
drive.mount('/content/drive')
```

# ***2. Importar Pandas y cargar archivos de Excel***


En este bloque aprendemos a cargar datos desde un formato de hoja de cálculo tradicional:
* `import pandas as pd`: Llamamos a nuestra herramienta principal de análisis de datos.
* `read_excel()`: A diferencia de `read_csv()`, esta función está diseñada específicamente para leer archivos de Microsoft Excel (`.xlsx`). Toma el archivo que tenemos guardado localmente y lo convierte en una tabla.
* `nba`: Al escribir únicamente el nombre de la variable al final de la celda, Colab nos imprime la tabla completa en la pantalla para poder verla.

```python
import pandas as pd
nba = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/udmy/Nba Teams.xlsx')
nba
```

# ***3. Resumen técnico general***


Una vez que la tabla está cargada, hacemos nuestro diagnóstico de rutina:
* `.info()`: Le aplica una "radiografía" a toda la tabla para conocer cuántas filas y columnas tiene en total, los nombres de los encabezados, la cantidad de datos válidos y el tipo de dato matemático de cada columna.

```python
nba.info()
```

# ***4. Vistazo rápido (Primeras filas)***


Si la tabla de la NBA es muy grande, imprimirla toda puede saturar nuestro cuaderno:
* `.head()`: Este método nos muestra por defecto únicamente las primeras 5 filas de la tabla. Es la herramienta ideal para confirmar visualmente que los datos están bien organizados sin ocupar tanto espacio en pantalla.

```python
nba.head()
```

# ***5. Combinar métodos (Encadenamiento)***


En Python podemos unir varias herramientas en una sola línea de código, ejecutándose de izquierda a derecha:
* `head(5).info()`: Aquí estamos "encadenando" dos métodos. Primero, el programa extrae únicamente las 5 primeras filas de la tabla (`.head(5)`). Inmediatamente después, le aplica la radiografía (`.info()`) de forma exclusiva a ese pequeño fragmento de 5 filas, ignorando el resto de la tabla original.

```python
nba.head(5).info()
```

# ***6. Resumen del proceso completo***


En conclusión, este archivo nos enseña el flujo de trabajo correcto para analizar archivos locales, desde la preparación del entorno hasta el diagnóstico:
* **1. Conexión a la nube:** Aprendimos que para usar documentos personales en Google Colab, el paso inicial obligatorio es "montar" nuestro Google Drive para darle permisos de lectura al programa.
* **2. Flexibilidad de lectura:** Comprobamos que Pandas es totalmente compatible con formatos empresariales tradicionales como Microsoft Excel (`.xlsx`), no solo con enlaces de internet.
* **3. Diagnóstico eficiente:** Reforzamos el uso de `.info()` y `.head()` para explorar la estructura de los datos sin saturar nuestro espacio visual.
* **4. Encadenamiento (Chaining):** Descubrimos que podemos combinar múltiples herramientas en una sola línea (como `.head().info()`) para aplicar análisis técnicos a fragmentos muy específicos de nuestra tabla.
