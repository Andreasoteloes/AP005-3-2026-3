# **2. Cargar los datos y optimizar tipos**


Importamos la librería y cargamos el archivo de Instagram aplicando las buenas prácticas de optimización que aprendimos antes:
* `read_csv()`: Cargamos el archivo desde la ruta exacta del Drive.
* `to_datetime()`: Convertimos las fechas a un formato real.
* `astype("category")` y `astype("bool")`: Transformamos las columnas categóricas y de tipo verdadero/falso para ahorrar memoria RAM y optimizar el rendimiento del equipo.

```python
import pandas as pd

insta = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/udmy/aplicada/Instagram.csv')
insta["reg date"] = pd.to_datetime(insta["reg date"])
insta["last login"] = pd.to_datetime(insta["last login"])
insta["gender"] = insta["gender"].astype("category")
insta["Country"] = insta["Country"].astype("category")
insta["Influencer"] = insta["Influencer"].astype("bool")
insta.info()
```

Comprobamos que la tabla esté lista para el filtrado:

# **4. Filtrar datos de forma directa (Condiciones booleanas)**


Para extraer únicamente las filas que cumplen con un requisito, colocamos la condición matemática dentro de los corchetes principales de la tabla (`insta[...]`):
* `insta[insta["gender"] == "Male"]`: Le pide a Pandas que filtre y devuelva solo las filas donde el género sea exactamente igual a "Male".
* `insta[insta["Country"] == "Australia"]`: Filtra la tabla para mostrar exclusivamente a los usuarios registrados en Australia.

```python
insta[insta["gender"] == "Male"]
insta[insta["Country"] == "Australia"]
```

# **5. El uso de "Máscaras" para filtrar**


Cuando las condiciones de filtrado son muy largas o complejas, los programadores suelen guardarlas primero en una variable intermedia llamada **máscara**:
* `mask = insta["gender"] == "Female"`: Creamos una regla lógica que evalúa celda por celda quiénes son mujeres (`True`) y quiénes no (`False`).
* `insta[mask]` / `insta[insta["Influencer"] == True]`: Aplicamos la máscara o la condición booleana directa para aislar a las usuarias mujeres o a los influencers confirmados.

```python
mask = insta["gender"] == "Female"
insta[mask]
insta[insta["Influencer"] == True]
```

# **6. Filtrar por negación (Excluir dato**s)


A veces no queremos buscar quién *sí* cumple una condición, sino exactamente lo contrario (quiénes quedan excluidos):
* `!=`: El operador de desigualdad (diferente de).
* `mask1 = insta["Country"] != "United States"`: Crea una regla para seleccionar a todos los países **excepto** Estados Unidos. Al pasar esta máscara a la tabla, filtramos y ocultamos todos los registros de ese país.

```python
mask1 = insta["Country"] != "United States"
insta[mask1]
```

# **7. Resumen del proceso completo**


En conclusión, este archivo nos enseña el mecanismo fundamental para extraer subconjuntos de datos basados en condiciones lógicas:
* **Filtros directos:** Aprendimos a aislar información escribiendo condiciones directamente dentro de los corchetes (ej. `== "Male"`).
* **El concepto de "Máscaras":** Comprendimos que guardar condiciones booleanas en variables intermedias (`mask`) hace que nuestro código sea mucho más ordenado, legible y fácil de mantener en proyectos grandes.
* **Exclusión de datos:** Descubrimos cómo usar el operador de desigualdad (`!=`) para limpiar o descartar registros específicos de nuestra base de datos.
