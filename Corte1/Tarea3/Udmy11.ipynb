# ***2. Cargar la tabla de datos***


Preparamos nuestro entorno leyendo el archivo original:
* `read_excel()`: Buscamos el archivo de la NBA usando su ruta exacta (path).
* `.head(10)`: Imprimimos las primeras 10 filas para visualizar en dónde tenemos datos faltantes (que Pandas muestra como `NaN` - Not a Number).

```python
import pandas as pd
nba = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/udmy/Nba Teams.xlsx')
nba.head(10)
```

# ***3. Rellenar huecos en toda la tabla (Método global)***


En lugar de borrar las filas con datos faltantes, podemos rellenarlas con un valor por defecto:
* `.fillna(0)`: Esta herramienta busca absolutamente todas las celdas vacías (`NaN`) en la tabla y las reemplaza por un número cero.
* **El detalle clave:** Como no le pusimos el parámetro `inplace = True`, esto es solo un simulacro o vista previa. Nuestra tabla original no se ha modificado permanentemente.

```python
nba.fillna(0)
```

# ***4. Rellenar huecos quirúrgicamente (Numérico)***


Si rellenamos toda la tabla con ceros, podríamos arruinar columnas de texto. Lo ideal es rellenar columna por columna:
* `nba[["College"]]`: Primero, usamos el método de extracción (con doble corchete) para aislar únicamente la columna de las universidades.
* `.fillna(0, inplace = True)`: Le encadenamos la herramienta de relleno. Ahora, si un jugador no tiene universidad registrada, se le asignará un "0", y gracias al `inplace = True`, este parche sí queda guardado permanentemente.

```python
nba[["College"]].fillna(0, inplace = True)
nba
```

# ***5. Rellenar huecos con etiquetas de texto***


El relleno no se limita a números. Si trabajamos con columnas de texto (como los nombres), podemos usar palabras:
* `nba["Name"]`: Extraemos la columna de nombres.
* `.fillna("Missing name", inplace = True)`: Si la base de datos olvidó registrar el nombre de un jugador, en lugar de poner un cero o dejarlo como `NaN`, el sistema escribirá automáticamente la frase "Missing name" (Nombre faltante), guardando el cambio para siempre.

```python
nba["Name"].fillna("Missing name", inplace = True)
nba
```

# ***6. Resumen del proceso completo***


En conclusión, este archivo nos da una alternativa menos destructiva para la limpieza de datos:
* **Conservación de información:** Aprendimos que `.fillna()` nos permite salvar filas que de otro modo tendríamos que borrar. Es como ponerle masilla a los huecos de nuestra base de datos.
* **Relleno global vs. específico:** Comprendimos que aplicar `.fillna()` a toda la tabla puede causar errores de formato (mezclar ceros numéricos en columnas de texto), por lo que la mejor práctica es aislar la columna que queremos reparar (ej. `nba["Name"]`).
* **Personalización de parches:** Comprobamos que podemos rellenar los datos vacíos tanto con valores numéricos (0) como con cadenas de texto ("Missing name"), adaptándonos a la naturaleza de cada columna.
