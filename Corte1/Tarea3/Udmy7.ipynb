# ***2. Cargar la tabla de datos***

Preparamos nuestro espacio de trabajo importando las herramientas y los datos:
* `import pandas as pd`: Traemos nuestra librería de análisis.
* `read_excel()`: Cargamos el archivo completo.

```python
import pandas as pd
nba = pd.read_excel("/content/drive/MyDrive/Colab Notebooks/udmy/Nba Teams.xlsx")
nba.head()
```

# ***3. Crear nuevas columnas (Asignación directa)***


En Pandas, si usamos corchetes con un nombre de columna que **no existe**, el programa crea esa columna automáticamente al final de la tabla.
* `nba["Teams"] = " "`: Crea una columna llamada "Teams" y rellena todas las filas con un espacio en blanco.
* `nba["nba Teams"] = " "`: Crea otra columna llamada "nba Teams" (con espacio en el nombre) y también la rellena con espacios.

```python
nba["Teams"] = " "
nba["nba Teams"] = " "
nba.head()
```

# ***5. Rellenar columnas con un valor fijo***


Aplicamos la misma lógica del Bloque 3, pero esta vez con texto real:
* `nba["Sports"] = "basket"`: Crea una nueva columna al final llamada "Sports" y le asigna la palabra "basket" a absolutamente todos los jugadores de la tabla.

```python
nba["Sports"] = "basket"
nba.head()
```

# ***6. Eliminar columnas permanentemente (Método Drop)***


Si creamos una columna por error o ya no la necesitamos, usamos la herramienta de eliminación:
* `.drop("Teams")`: Busca la columna que se llama exactamente así para borrarla.
* `axis = 1`: Es una regla de Pandas. El "1" significa que vamos a borrar una columna (vertical). Si pusiéramos "0", borraría una fila (horizontal).
* `inplace = True`: Es el equivalente al botón de "Guardar cambios". Le dice a Pandas que aplique esta eliminación de forma permanente en nuestra tabla original.

```python
nba.drop("Teams", axis = 1, inplace = True)
#nba.insert(2, column = "Teams", value = "Colombians")
#nba.insert(2, column = "Sportss", value = "Turmequé")
nba
```

# ***7. Insertar columnas en posiciones exactas***


El método de corchetes que vimos arriba siempre pone las columnas nuevas al final de la tabla. Si queremos ponerla en medio, usamos esto:
* `.insert(2, ...)`: La herramienta de inserción nos pide primero la posición numérica. Como Python cuenta desde el 0 (0, 1, 2), el número `2` significa que la columna aparecerá en la tercera posición.
* `column = "Teams"`: Define el título que tendrá la nueva columna.
* `value = "Colombians"`: Define el texto con el que se van a rellenar todas las filas.

```python
nba.insert(2, column = "Teams", value = "Colombians")
nba
```
