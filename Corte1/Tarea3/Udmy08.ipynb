# ***2. Cargar la tabla de datos***


Importamos nuestra herramienta y cargamos la información original:
* `import pandas as pd`: Traemos a Pandas.
* `read_excel()`: Cargamos el archivo con los datos completos de los equipos.
* `.head()`: Visualizamos el encabezado para confirmar que tenemos todas las columnas originales (incluyendo "Salary").

```python
import pandas as pd
nba = pd.read_excel("/content/drive/MyDrive/Colab Notebooks/udmy/Nba Teams.xlsx")
nba.head()
```

# ***3. Eliminar una columna (Visualización temporal)***


Usamos el método `.drop()` para borrar la columna de salarios:
* `.drop("Salary")`: Le indicamos que queremos deshacernos de la columna con ese nombre exacto.
* `axis = 1`: Le recordamos a Pandas que estamos apuntando a una columna (vertical).
* **El truco de esta línea:** Como no usamos el parámetro `inplace = True`, Pandas solo nos mostrará un "simulacro" o vista previa de cómo se vería la tabla sin esa columna, pero no modificará el archivo real.

```python
nba.drop("Salary", axis = 1)
```

# ***4. Comprobar que la tabla no ha cambiado***


Si llamamos de nuevo a nuestra tabla escribiendo simplemente su nombre, veremos la prueba de lo que explicamos arriba:
* Al imprimir `nba`, notaremos que la columna "Salary" sigue ahí intacta. Como fue un borrado temporal, la estructura original de nuestra memoria está a salvo.

```python
nba
```

# ***5. Eliminar permanentemente (Guardar cambios)***


Ahora sí aplicaremos la eliminación real:
* `inplace = True`: Al agregar este parámetro, le estamos dando la orden a Pandas de que "guarde los cambios". Ahora la destrucción de la columna "Salary" sí se aplica sobre nuestra tabla original de manera irreversible (dentro de esta sesión).

```python
nba.drop("Salary", axis = 1, inplace = True)
```

# ***6. Borrado masivo (Eliminar varias columnas)***


Si necesitamos borrar mucha información de golpe, no hace falta escribir `.drop()` muchas veces:
* `["Team", "Division"]`: Al igual que cuando extraíamos columnas, si le entregamos a la herramienta una **lista de Python `[ ]`**, podemos pedirle que borre varias columnas simultáneamente.
* Con `axis = 1` confirmamos que son columnas y con `inplace = True` guardamos esta limpieza masiva de forma permanente.

```python
nba.drop(["Team", "Division"], axis = 1, inplace = True)
```

# ***7. Resumen del proceso completo***


En conclusión, este archivo nos enseña el manejo seguro de la eliminación de datos, un paso crítico en cualquier limpieza de bases de datos:
* **Simulacros vs. Cambios Reales:** El aprendizaje más importante es el uso del parámetro `inplace = True`. Sin él, cualquier alteración que hagamos a la tabla será solo una vista previa temporal.
* **Manejo de Listas:** Reforzamos que en Pandas, siempre que queramos aplicar una acción (como extraer o borrar) a más de un elemento, debemos agrupar sus nombres dentro de una lista de Python `[ ]`.
