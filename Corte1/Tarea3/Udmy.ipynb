# ***1. Cargar y visualizar los datos***


En este paso traemos la información desde internet a nuestro cuaderno de trabajo:
* `import pandas as pd`: Llama a nuestra herramienta principal de análisis de datos.
* `read_csv()`: Toma el enlace web (la URL) y convierte ese texto en una tabla manejable.
* `.head()`: Muestra únicamente las primeras 5 filas para comprobar que la tabla cargó bien, sin saturar la pantalla.

```python
import pandas as pd
url = "https://data.cityofnewyork.us/resource/rc75-m7u3.csv"
covid = pd.read_csv(url)
covid.head()
```

# ***2. Resumen de la estructura interna***


Antes de analizar, necesitamos saber con qué estamos trabajando usando este método de diagnóstico:
* `.info()`: Funciona como una radiografía. Nos indica la cantidad total de columnas y filas, los nombres de los encabezados, los datos faltantes y qué tipo de dato matemático contiene cada columna (textos, números, etc.).

```python
covid.info()
```

# ***3. Guardar el archivo completo***

Como los datos solo están en la memoria temporal del programa, necesitamos exportarlos físicamente:
* `.to_csv()`: Es la función de "Guardar como". Toma toda nuestra tabla y crea un archivo nuevo llamado `NYC_covid_cases.csv` que se guardará en nuestra carpeta.

```python
covid.to_csv('NYC_covid_cases.csv')
```

# ***4. Exportar columnas específicas***

Para no crear un archivo pesado con información innecesaria, podemos hacer una exportación selectiva:
* `columns = [ ]`: Usamos de nuevo `.to_csv()`, pero le agregamos este parámetro. Le entregamos una lista en corchetes con los tres nombres exactos de las columnas que queremos salvar, descartando el resto.

```python
covid.to_csv('NYC1_covid_cases.csv', columns = ['case_count', 'hospitalized_count', 'death_count'])
```

# ***5. Resumen del proceso completo***


En conclusión, este primer archivo demuestra el ciclo básico inicial de cualquier análisis de datos usando Pandas. El proceso fluye en tres pasos clave:
* **1. Extracción:** Tomamos datos "crudos" alojados en internet y los convertimos en una tabla de trabajo.
* **2. Exploración:** Le hacemos un chequeo médico rápido a la tabla para conocer su tamaño, qué tipo de datos contiene y si tiene errores o vacíos.
* **3. Exportación:** Guardamos copias físicas de seguridad en nuestra carpeta, ya sea con la información completa o filtrando solo las columnas útiles para limpiar nuestro espacio de trabajo.
