# **2. Cargar datos y convertir a tipo Categoría**


* `emp.nunique()`: Muestra cuántos valores únicos tiene cada columna (ideal para saber qué columnas repiten mucho sus datos).
* `.astype("category")`: Convierte columnas de texto repetitivo en datos "categóricos". Esto reduce drásticamente el consumo de memoria RAM en computadoras y acelera los cálculos en tablas gigantes.

```python
import pandas as pd
emp = pd.read_csv("/content/drive/MyDrive/aplicada/Employee.csv")
emp.head()
emp.info()
emp.nunique()
emp[["department", "position"]] = emp[["department", "position"]].astype("category")
emp.info()
```

# **3. Formateo y limpieza de textos**


Para modificar cadenas de caracteres usamos funciones de texto:
* `.upper()` / `.lower()`: Vuelven todo el texto a mayúsculas o minúsculas.
* `.str`: Es el acceso especial de Pandas para aplicar herramientas de texto a columnas enteras.
* `.title()`: Convierte una cadena de texto para que la primera letra de cada palabra sea mayúscula y el resto minúscula (formato título).

```python
"happy".upper()
"PYTHON".lower()

emp["name"]
emp["name"].str.lower().str.title()
```

# **4. Resumen del proceso completo**


* **Optimización:** Aprendimos a usar `.astype("category")` para ahorrar memoria en columnas con textos repetitivos.
* **Manipulación de Strings:** Descubrimos cómo usar `.str` junto con métodos como `.lower()` y `.title()` para limpiar y estandarizar la presentación de los nombres en toda la tabla.
