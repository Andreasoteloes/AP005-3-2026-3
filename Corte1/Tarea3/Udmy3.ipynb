# ***1. Preparar el entorno***


Igual que en los archivos anteriores, el primer paso indispensable es preparar nuestro espacio de trabajo:
* `import pandas as pd`: Llamamos a la herramienta principal de análisis de datos y le asignamos su apodo clásico (`pd`) para facilitar su uso.

```python
import pandas as pd
```

# ***2. Crear datos estructurados (Diccionario)***


En el archivo pasado usamos "listas de listas", pero aquí usaremos una estructura mucho más eficiente: los diccionarios.
* `data = { ... }`: Construimos un diccionario. La magia de hacerlo así es que las "claves" (como "name" o "age") funcionarán como los encabezados de las columnas, y los "valores" (las listas `[ ]` de adentro) contendrán la información de cada fila. Esta es la forma más común de armar datos manualmente.

```python
data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}
```

# ***3. Transformar el Diccionario en Tabla***


Ahora le entregamos nuestro diccionario a Pandas para que construya la tabla visual:
* `pd.DataFrame()`: Toma el diccionario y lo transforma en tabla. La gran ventaja aquí es que Pandas reconoce automáticamente las "claves" del diccionario y las usa como nombres de las columnas, ahorrándonos el paso de usar el parámetro `columns`.
* `index = [ ]`: Aprovechamos para cambiar la numeración automática de las filas (0, 1, 2) por etiquetas descriptivas personalizadas ("Kid1", "Kid2", "Kid3").

```python
pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])
```

# ***4. Resumen del proceso completo***


En conclusión, este archivo nos enseña la forma más eficiente de crear tablas manualmente en Pandas:
* **Estructura ideal:** Comprobamos que usar Diccionarios (en lugar de listas anidadas) es una mejor práctica porque nos permite bautizar las columnas directamente desde la creación de los datos.
* **Automatización:** Al usar `pd.DataFrame()`, la librería es lo suficientemente inteligente para entender la anatomía del diccionario sin que tengamos que explicarle cuál es la columna y cuál es el dato.
* **Personalización de filas:** Mantenemos la capacidad de usar el parámetro `index` para asignar identificadores únicos a cada fila, mejorando la lectura de la tabla.
