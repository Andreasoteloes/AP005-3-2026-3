# **2. Cargar los datos y optimizar tipos de datos desde la lectura**


Aplicamos una optimización avanzada al cargar nuestro archivo:
* `parse_dates = [...]`: Este parámetro especial dentro de `read_csv` le indica a Pandas que convierta automáticamente esas columnas en formato de fechas reales desde el mismo momento en que abre el archivo, ahorrándonos líneas de código posteriores.
* `astype(...)`: Convertimos las columnas categóricas y booleanas para optimizar el consumo de memoria RAM de la computadora.

```python
import pandas as pd

insta = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/udmy/aplicada/Instagram.csv', parse_dates = ["reg date", "last login"])
insta["gender"] = insta["gender"].astype("category")
insta["Country"] = insta["Country"].astype("category")
insta["Influencer"] = insta["Influencer"].astype("bool")

insta.head()
```

# **3. Aplicar múltiples condiciones simultáneas (Operador AND)**


Si queremos filtrar filas que cumplan **todas** las reglas al mismo tiempo, usamos el operador ampersand (`&`), que representa la conjunción lógica "Y":
* `(Condición 1) & (Condición 2) & (Condición 3)`: Es una **regla de oro en Pandas** obligatoria: cada condición independiente debe ir obligatoriamente encerrada entre paréntesis `( )`. Si olvidas los paréntesis, Python se confundirá y arrojará un error matemático.
* En este ejemplo, filtramos usuarios que sean hombres (`"Male"`), cuya fecha de registro sea posterior al 31 de diciembre de 2015 (`> "2015-12-31"`), y que vivan en Estados Unidos (`"United States"`).

```python
mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")
insta[mask]
```

# **4. Organizar condiciones complejas usando variables (Máscaras múltiples)**


Cuando las consultas son muy largas, escribir todo en una sola línea puede volverse confuso de leer. Una excelente práctica profesional es dividir la lógica en partes:
* `mask1`, `mask2`, `mask3`: Guardamos cada regla de filtrado en una variable independiente.
* `insta[mask1 & mask2 & mask3]`: Al final, unimos todas las máscaras dentro de los corchetes usando el operador `&`. Esto hace que el código sea sumamente limpio, ordenado y fácil de depurar si algo falla.

```python
mask1 = insta["gender"] == "Male"
mask2 = insta["reg date"] > "2015-12-31"
mask3 = insta["Country"] == "United States"
insta[mask1 & mask2 & mask3]
```

# **5. Cuidado con las mayúsculas y minúsculas**


Un error común en programación es asumir que las palabras son iguales sin importar cómo se escriban:
* `(insta["gender"] == "male")`: Si escribimos la palabra en minúscula cuando en la base de datos está registrada con mayúscula inicial (`"Male"`), Pandas no encontrará ninguna coincidencia y nos devolverá una tabla completamente vacía. ¡Siempre debemos verificar el texto exacto!

```python
mask = (insta["gender"] == "male") & (insta["reg date"] > "2015-12-31")
insta[mask]
```
